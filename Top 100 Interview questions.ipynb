```sql
-- 1. Find duplicate records in a table (Amazon)
-- Using bakehouse sales data to find duplicate transactions
SELECT customerID, product, COUNT(*) as duplicate_count
FROM samples.bakehouse.sales_transactions
GROUP BY customerID, product
HAVING COUNT(*) > 1
ORDER BY duplicate_count DESC;
```

```sql
-- 2. Retrieve the second highest price from sales transactions
-- Demonstrating the second-highest pattern using unitPrice
SELECT MAX(unitPrice) as SecondHighestPrice
FROM samples.bakehouse.sales_transactions
WHERE unitPrice < (SELECT MAX(unitPrice) FROM samples.bakehouse.sales_transactions);
```

```sql
-- 2b. Alternative: Second highest using DENSE_RANK (more robust)
SELECT DISTINCT totalPrice
FROM (
  SELECT totalPrice,
         DENSE_RANK() OVER (ORDER BY totalPrice DESC) as rank
  FROM samples.bakehouse.sales_transactions
)
WHERE rank = 2;
```

```sql
-- 3. Find transactions without franchise match (LEFT JOIN pattern)
-- Using sales transactions and franchises to demonstrate LEFT JOIN
SELECT t.transactionID, t.customerID, t.franchiseID, t.product
FROM samples.bakehouse.sales_transactions t
LEFT JOIN samples.bakehouse.sales_franchises f ON t.franchiseID = f.franchiseID
WHERE f.franchiseID IS NULL
LIMIT 10;
```

```sql
-- 4. Calculate the total revenue per product (PayPal)
-- Using bakehouse sales transactions
SELECT product, 
       SUM(quantity * unitPrice) AS total_revenue,
       COUNT(*) as transaction_count,
       SUM(quantity) as total_quantity
FROM samples.bakehouse.sales_transactions
GROUP BY product
ORDER BY total_revenue DESC;
```
