```sql
%sql

-- =====================================================
-- Q1: Total Oil & Gas Production by Field (Active Wells)
-- =====================================================
SELECT
    Field,
    SUM(OilProduced_bbl) AS Total_OilProduced_bbl,
    SUM(GasProduced_mcf) AS Total_GasProduced_mcf
FROM well_production
WHERE Status = 'Active'
GROUP BY Field;
```

```sql
%sql

-- =====================================================
-- Q1: Total Oil & Gas Production by Field (Active Wells)
-- =====================================================
SELECT
    Field,
    SUM(OilProduced_bbl) AS Total_OilProduced_bbl,
    SUM(GasProduced_mcf) AS Total_GasProduced_mcf
FROM well_production
WHERE Status = 'Active'
GROUP BY Field;
```

```sql
-- =====================================================
-- Q2: Average Operating Cost by Operator
-- =====================================================
SELECT
    OperatorName,
    AVG(OperatingCost) AS Avg_OperatingCost
FROM well_production
GROUP BY OperatorName
ORDER BY Avg_OperatingCost DESC;
```

```sql
-- =====================================================
-- Q3: Number of Wells per Region (More than 2 Wells)
-- =====================================================
SELECT
    Region,
    COUNT(*) AS Total_Wells
FROM well_production
GROUP BY Region
HAVING COUNT(*) > 2;
```

```sql
-- =====================================================
-- Q4: Top 3 Fields by Total Oil Production
-- =====================================================
SELECT
    Field,
    SUM(OilProduced_bbl) AS Total_OilProduced_bbl
FROM well_production
GROUP BY Field
ORDER BY Total_OilProduced_bbl DESC
LIMIT 3;
```

```sql
-- =====================================================
-- Q5: Minimum & Maximum Depth by Well Type
-- (Excluding Abandoned Wells)
-- =====================================================
SELECT
    WellType,
    MIN(DepthMeters) AS Min_DepthMeters,
    MAX(DepthMeters) AS Max_DepthMeters
FROM well_production
WHERE Status <> 'Abandoned'
GROUP BY WellType;
```

```sql
-- =====================================================
-- Q6: Top 5 Operators by Average Oil Production
-- =====================================================
SELECT
    OperatorName,
    COUNT(*) AS Total_Wells,
    AVG(OilProduced_bbl) AS Avg_OilProduced_bbl
FROM well_production
GROUP BY OperatorName
ORDER BY Avg_OilProduced_bbl DESC
LIMIT 5;
```
