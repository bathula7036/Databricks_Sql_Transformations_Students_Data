```sql
%sql
-- COMMAND ----------
-- Step 1: Create the table and load sample data
CREATE OR REPLACE TABLE customers (
  CustomerID   INT,
  CustomerName STRING,
  City         STRING,
  AccountType  STRING,
  Balance      INT,
  LoanAmount   INT
);

INSERT INTO customers VALUES
(101, 'John',   'Dallas', 'Savings', 5000, 10000),
(102, 'Mary',   'Austin', 'Current', 2500, 5000),
(103, 'David',  'Dallas', 'Savings', 8000, 0),
(104, 'Emma',   'Houston','Savings', 1500, 3000),
(105, 'Alex',   'Austin', 'Current', 7000, 12000),
(106, 'Sophia', 'Dallas', 'Savings', 5000, 8000);
```

```sql
%sql
select * from customers;
```

```sql
%sql
-- COMMAND ----------
-- Step 2: Display CustomerName, City, Balance, and LoanAmount
SELECT CustomerName, City, Balance, LoanAmount
FROM customers;
```

```sql
%sql
-- COMMAND ----------
-- Step 3: Display unique cities
SELECT DISTINCT City
FROM customers;
```

```sql
%sql
-- COMMAND ----------
-- Step 4: Sort customers by LoanAmount (Highest to Lowest)
SELECT CustomerName, City, Balance, LoanAmount
FROM customers
ORDER BY LoanAmount ASC;
```

```sql
%sql
-- COMMAND ----------
-- Step 5: Show only the top 5 customers based on LoanAmount
SELECT CustomerName, City, Balance, LoanAmount
FROM customers
ORDER BY LoanAmount DESC
LIMIT 10;
```

```sql
%sql
-- COMMAND ----------
-- Step 6: Summary statistics
SELECT
  COUNT(CustomerID)   AS Total_Customers,
  SUM(LoanAmount)     AS Total_Loan_Amount,
  MIN(Balance)        AS Min_Balance,
  MAX(Balance)        AS Max_Balance,
  AVG(Balance)         AS Avg_Balance
FROM customers;
```
