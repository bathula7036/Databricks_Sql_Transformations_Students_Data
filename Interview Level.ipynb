```sql
-- ==========================================================
-- SQL Interview Questions (Amazon, Google, Uber, Walmart...)
-- ==========================================================
--------------------------------------------------------------
-- Q1. Find Duplicate Records (Amazon)
--------------------------------------------------------------

%sql
SELECT
    Email,
    COUNT(*) AS Duplicate_Count
FROM employees
GROUP BY Email
HAVING COUNT(*) > 1;

```

```sql
--------------------------------------------------------------
-- Q2. Retrieve Second Highest Salary
--------------------------------------------------------------
%sql
SELECT MAX(Salary) AS Second_Highest_Salary
FROM employees
WHERE Salary < (
    SELECT MAX(Salary)
    FROM employees
);
```

```sql
--------------------------------------------------------------
-- Q3. Find Employees Without Department (Uber)
--------------------------------------------------------------
%sql
SELECT
    e.EmployeeID,
    e.Name
FROM employees e
LEFT JOIN departments d
ON e.DepartmentID = d.DepartmentID
WHERE d.DepartmentID IS NULL;
```

```sql
--------------------------------------------------------------
-- Q4. Calculate Total Revenue Per Product (PayPal)
--------------------------------------------------------------
%sql
SELECT
    ProductID,
    SUM(Quantity * Price) AS Total_Revenue
FROM sales
GROUP BY ProductID;
```

```sql
--------------------------------------------------------------
-- Q5. Top 3 Highest Paid Employees (Google)
--------------------------------------------------------------
%sql
SELECT
    EmployeeID,
    Name,
    Salary
FROM employees
ORDER BY Salary DESC
LIMIT 3;
```

```sql
--------------------------------------------------------------
-- Q6. Customers Who Purchased but Never Returned Products
-- (Walmart)
--------------------------------------------------------------
%sql
SELECT DISTINCT
    o.CustomerID
FROM orders o
LEFT JOIN returns r
ON o.OrderID = r.OrderID
WHERE r.OrderID IS NULL;
```
